**Key areas at a glance:**

| Area | Core Question | Key Techniques |
|------|--------------|----------------|
| **Mechanistic Interpretability** | What algorithms do models learn? | Circuits analysis, feature identification, superposition theory |
| **Sparse Autoencoders** | How do we extract interpretable features from superposed representations? | Dictionary learning, SAE training, feature dashboards |
| **Causal Methods** | Which components cause specific behaviors? | Activation patching, path patching, causal tracing |
| **Representation Analysis** | What information is encoded where? | Probing classifiers, logit lens, tuned lens |
| **Representation Engineering** | Can we control model behavior via representations? | Steering vectors, activation addition, contrastive activations |
| **Circuit Tracing** | Can we trace full computational graphs through a model? | Attribution graphs, cross-layer tracing via SAE features |

## How the Self-Updating System Works

This study guide includes a lightweight research tracker that keeps you current with new interpretability work.

**How it works:**

- The `research_tracker/` directory contains scripts that periodically fetch new papers from **arXiv** and **Semantic Scholar** matching interpretability-related queries.
- Fetched results are processed and summarized into digest files.
- Generated digests appear in the `digests/` directory, organized by date.
- Each digest includes paper titles, abstracts, relevance scores, and links -- so you can quickly scan what is new without leaving your notebook workflow.

You do not need to run the tracker to use the study guide notebooks. It is an optional tool for staying up to date as the field moves quickly.
